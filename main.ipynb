# ***Task 1.*** 

 > 1.a Load the data-set in notebook. Each name is separated using a new line.

 > 1.b Create Employee ID: Starting from 2020001. Example 2020001, 2020002 and so on for 1000 employee.

 > 1.c Email ID: first_name.last_name@edureka.co.

 >1.d Password: It should be an alphanumeric value, must having capital letters, small letters, and special symbols.

```python
#---Importing required libraries---#
import pandas as pd
import numpy as np
import random
```

```python
import urllib.request

file = "https://www.dropbox.com/s/onl5ac2ea3v11aw/names.txt?dl=1"

urllib.request.urlretrieve(file, "names.txt")

print("Dataset downloaded successfully!")
```

```python
from os import read, write, remove
files = open('names.txt', 'r')
names = files.read()
files.close()
```

```python
#---Read the names from the dataset---#
print(names)
```

```python
#---Read the type, shape, & size of dataset---#

#---Read the type of dataset---#
print('The type of the dataset is:', type(names))

#---Read the shape of dataset---#
print('The shape of the dataset is:', np.shape(names))

#---Read the size of dataset---#
print('The size of the dataset is:', np.size(names))

#---Read the length of dataset---#
print('The length of the dataset is:', len(names))
```

```python
#---Split the names into a list---#
names = names.strip().split('\n')
print(names[0:10])
print('The length of the dataset is:', len(names))
```

```python
#---Password Generator---#
sp_symbols = ['!', '@', '#', '$', '&', '?']
def pass_generator():

    pwd = []
    caps_alpha = [chr(random.randrange(65, 65 + 26)) for _ in range(random.randrange(1, 4))]
    small_alpha = [chr(random.randrange(97, 97 + 26)) for _ in range(random.randrange(3, 6))]
    np.random.shuffle(sp_symbols)
    pwd.append(sp_symbols[0])
    num = [str(random.randrange(1, 10)) for _ in range(random.randrange(1, 4))]
    pwd = num + caps_alpha + small_alpha + pwd
    np.random.shuffle(pwd)
    pwd = ''.join(pwd)
    return pwd
```

```python
pass_generator()
```

```python
#---Email ID Generator---#
def email_id(name):
    name = name.split()
    email = '.'.join(name)
    return email.lower() + '@edureka.co'
```

```python
print(email_id('Edward Davis'))
```

```python
emp_id = [id for id in range(2020001, 2020001 + 1000)]
email_id = [email_id(name) for name in (names)]
password = [pass_generator() for _ in range(1000)]
```

```python
print(emp_id[:10])
print(email_id[:10])
print(password[:10])
```
